## 1. Define the sample record

We first define the discrete-time record and the sinusoid parameters.

### Teaching point

A one-second record sampled at \(F_s=1000\) samples/s contains

\[
N = F_s T = 1000
\]

samples.

The integer sample index is

\[
n = 0,1,\ldots,N-1
\]

and the corresponding sample times are

\[
t[n] = 
rac{n}{F_s}.
\]

Because Python and C/C++ both use zero-based indexing, `n[0]` is the first sample index and `t[0]` is the first sample time.

### Useful live checks

After running the cell, you can optionally type these into a new code cell:

```python
%whos
```

```python
n[:10]
```

```python
t[:10]
```

The last two commands let students see exactly how the integer sample index becomes a time axis.

````python
# -------------------------------------------------------------------------
# SECTION 1: DEFINE THE SAMPLE RECORD
# -------------------------------------------------------------------------
# The goal is to create the discrete sample indices n and corresponding
# sample times t for a one-second record.

# Sampling frequency Fs = 1000 samples/second.
# This means adjacent samples are separated by 1/Fs = 0.001 seconds.
Fs = 1000              # Sampling rate in samples per second

# Total record duration in seconds.
T = 1                  # Record duration in seconds

# Number of samples in the record.
# Fs*T is 1000*1 = 1000.
#
# int(...) explicitly converts the result to Python's integer type.
# In C/C++ terms, think of this as making sure N is an integer-sized count.
N = int(Fs * T)        # Number of samples

# np.arange(N) creates the integer sequence
#     [0, 1, 2, ..., N-1]
# so here n has 1000 elements: 0 through 999.
#
# This is the discrete-time sample-index vector.
n = np.arange(N)       # Sample indices

# NumPy performs element-by-element division over the entire array.
# Therefore t contains
#     [0/Fs, 1/Fs, 2/Fs, ..., (N-1)/Fs]
#   = [0, 0.001, 0.002, ..., 0.999] seconds.
#
# In C/C++ this would normally require a loop over all N elements.
t = n / Fs             # Sample times

# -------------------------------------------------------------------------
# SINUSOID PARAMETERS
# -------------------------------------------------------------------------
# Peak amplitude of the deterministic cosine.
A = 1                  # Sinusoid amplitude

# Sinusoid frequency. A 50 Hz tone completes 50 cycles per second.
f0 = 50                # Sinusoid frequency in Hz

# Starting phase. Zero radians means the cosine begins at its maximum value.
phi = 0                # Initial phase in radians

# -------------------------------------------------------------------------
# PRINT A FEW SANITY CHECKS
# -------------------------------------------------------------------------
# f"..." is a Python formatted string literal, similar in purpose to printf.
#
# {N:d} formats N as a decimal integer.
print(f"Number of samples: {N:d}")

# {T:.2f} formats T as fixed-point with two digits after the decimal.
print(f"Record duration: {T:.2f} seconds")

# f0*T gives the number of sinusoidal cycles in the one-second record.
# {:.0f} prints the result with zero digits after the decimal point.
print(f"Number of sinusoidal cycles: {f0*T:.0f}")

# Optional live inspection in a new cell:
#     %whos
#     n[:10]
#     t[:10]
#
# Python slicing uses a half-open range. n[:10] means elements 0 through 9.
````

## 2. Generate a deterministic sinusoid twice

### Prediction

If all parameters are unchanged, should the two sinusoid records match exactly?

The signal is

\[
x[n] = A\cos\left(2\pi f_0 t[n]+\phi
ight).
\]

Because this is deterministic, the same input parameters must produce the same samples every time.

### C/C++ mental model

The vectorized NumPy expression

```python
A * np.cos(2 * np.pi * f0 * t + phi)
```

is conceptually equivalent to a C++ loop that evaluates `cos(...)` once for every element of `t`.

### What to point out

The plot shows only the first 100 samples. At 1000 samples/s, that is 0.1 seconds. A 50 Hz sinusoid completes 5 cycles in 0.1 seconds.

````python
# -------------------------------------------------------------------------
# SECTION 2: GENERATE THE SAME DETERMINISTIC SINUSOID TWICE
# -------------------------------------------------------------------------
# Mathematical model:
#     x[n] = A cos(2*pi*f0*t[n] + phi)
#
# np.pi is pi.
# Because t is a NumPy array, every arithmetic operation below is vectorized.
# np.cos(...) evaluates cosine once for every element of the phase array.
#
# x1 therefore contains N floating-point sinusoid samples.
x1 = A * np.cos(2 * np.pi * f0 * t + phi)

# Repeat the exact same deterministic calculation with the exact same inputs.
# There is no random component, so x2 should contain exactly the same samples.
x2 = A * np.cos(2 * np.pi * f0 * t + phi)

# -------------------------------------------------------------------------
# VERIFY THAT THE TWO ARRAYS ARE IDENTICAL
# -------------------------------------------------------------------------
# np.array_equal(a,b) returns Python True only if the two arrays have the same
# shape and every corresponding element compares equal.
sinusoids_are_identical = np.array_equal(x1, x2)

# x1 - x2 subtracts corresponding elements.
# np.abs(...) takes the magnitude/absolute value of every difference.
# np.max(...) returns the largest element.
#
# For two identical arrays, the largest absolute sample difference is zero.
largest_difference = np.max(np.abs(x1 - x2))

# bool values in Python are True/False.
# int(True) is 1 and int(False) is 0, so this printout uses 1/0.
print(f"Are the two sinusoid records identical? {int(sinusoids_are_identical)}")

# :.3e requests scientific notation with three digits after the decimal.
print(f"Largest sample difference: {largest_difference:.3e}")

# -------------------------------------------------------------------------
# PLOT THE FIRST 100 SAMPLES
# -------------------------------------------------------------------------
# Create a new figure.
plt.figure()

# t[:100] and x1[:100] select samples 0 through 99.
#
# 100 samples / 1000 samples/s = 0.1 seconds.
# At 50 Hz, 0.1 seconds contains 5 complete cycles.
#
# linewidth=1.5 simply makes the curve easier to see.
plt.plot(t[:100], x1[:100], linewidth=1.5)

# Axis labels and title.
plt.xlabel("Time (seconds)")
plt.ylabel("Amplitude")
plt.title("Deterministic Sinusoid")

# Render the figure in the notebook.
plt.show()
````

## 3. Generate two realizations of Gaussian noise

### Prediction

Should two records generated from the same Gaussian-noise model contain identical samples?

No. They share the same **probability model**, but each call to the random generator produces a different **realization**.

We want zero-mean Gaussian noise with variance

\[
\sigma_w^2 = 0.05.
\]

`rng.standard_normal(N)` generates \(N\) samples with mean 0 and variance 1. Multiplying those samples by \(\sqrt{0.05}\) gives variance 0.05 because

\[
\mathrm{Var}(aX)=a^2\mathrm{Var}(X).
\]

### Vocabulary to emphasize

- **Model/distribution:** the rule governing the random samples.
- **Realization:** one particular array of random samples generated from that model.
- `w1` and `w2` are different realizations of the same Gaussian-noise model.

````python
# -------------------------------------------------------------------------
# SECTION 3: GENERATE TWO GAUSSIAN-NOISE REALIZATIONS
# -------------------------------------------------------------------------

# Desired variance of the Gaussian noise process.
# The model is zero mean with variance 0.05.
noise_variance = 0.05

# Standard deviation is the square root of variance.
# sqrt(0.05) is approximately 0.2236.
noise_std = np.sqrt(noise_variance)

# rng.standard_normal(N) generates N independent samples from a standard
# normal distribution:
#     mean = 0
#     variance = 1
#
# Multiplying by noise_std scales the variance:
#     Var(aX) = a^2 Var(X)
# so Var(noise_std * X) = noise_variance.
#
# w1 is one finite realization of the Gaussian-noise model.
w1 = noise_std * rng.standard_normal(N)

# Calling the generator again advances its internal state and produces
# another set of samples. w2 obeys the same statistical model as w1 but is
# a different realization.
w2 = noise_std * rng.standard_normal(N)

# -------------------------------------------------------------------------
# VERIFY THAT DIFFERENT REALIZATIONS ARE NOT SAMPLE-BY-SAMPLE IDENTICAL
# -------------------------------------------------------------------------
noise_records_are_identical = np.array_equal(w1, w2)

# Find the largest absolute sample difference between the two realizations.
largest_noise_difference = np.max(np.abs(w1 - w2))

# Expect 0 here, meaning False.
print(f"Are the two noise records identical? {int(noise_records_are_identical)}")

# The actual largest difference is realization-dependent.
print(f"Largest sample difference: {largest_noise_difference:.4f}")

# -------------------------------------------------------------------------
# DISPLAY THE FIRST 50 SAMPLES OF BOTH REALIZATIONS
# -------------------------------------------------------------------------
# plt.subplots(2,1,...) creates a figure containing 2 rows and 1 column of axes.
#
# fig is the overall Figure object.
# ax is an array containing the two Axes objects:
#     ax[0] = upper plot
#     ax[1] = lower plot
#
# sharex=True tells both plots to use the same horizontal axis scale.
fig, ax = plt.subplots(2, 1, figsize=(8, 5), sharex=True)

# A stem plot is useful for emphasizing that these are discrete-time samples.
# n[:50] supplies sample indices 0 through 49.
# w1[:50] supplies the corresponding 50 noise values.
ax[0].stem(n[:50], w1[:50])
ax[0].set_ylabel("Amplitude")
ax[0].set_title("Noise Realization 1")

# Plot the second realization using the lower axes object.
ax[1].stem(n[:50], w2[:50])
ax[1].set_xlabel("Sample index n")
ax[1].set_ylabel("Amplitude")
ax[1].set_title("Noise Realization 2")

# Adjust spacing so labels do not overlap.
plt.tight_layout()

# Render the figure in Jupyter.
plt.show()
````

## 4. Calculate sample statistics

We now estimate the mean and variance from each finite noise realization.

### Mean

For a finite record,

\[
ar{w}=
rac{1}{N}\sum_{n=0}^{N-1}w[n].
\]

The theoretical Gaussian mean is zero, but a finite realization will usually have a sample mean that is close to zero rather than exactly zero.

### Variance and `ddof=0`

NumPy computes variance using a denominator of

\[
N-	exttt{ddof}.
\]

Therefore,

```python
np.var(w1, ddof=0)
```

uses the denominator \(N\).

`ddof` means **delta degrees of freedom**. Setting it to zero explicitly tells NumPy not to subtract any degrees of freedom from \(N\).

This notebook intentionally uses the population/signal-processing convention

\[

rac{1}{N}\sum_{n=0}^{N-1}(w[n]-ar{w})^2
\]

rather than the common unbiased statistical estimator that divides by \(N-1\).

### Teaching point

The specified model has variance 0.05. The two measured variances should be near 0.05, but they will generally not be identical to each other or exactly 0.05 because each estimate comes from only 1000 random samples.

````python
# -------------------------------------------------------------------------
# SECTION 4: SAMPLE MEAN AND SAMPLE VARIANCE
# -------------------------------------------------------------------------
# We now estimate statistics from the two finite realizations.
#
# Important teaching distinction:
#     The MODEL mean is exactly 0.
#     The SAMPLE mean from 1000 random values will generally be near 0,
#     but not exactly 0.

# np.mean(...) sums all elements and divides by the number of elements.
mean_w1 = np.mean(w1)
mean_w2 = np.mean(w2)

# -------------------------------------------------------------------------
# VARIANCE AND ddof
# -------------------------------------------------------------------------
# NumPy's variance denominator can be thought of as:
#     N - ddof
#
# Therefore ddof=0 means:
#     denominator = N - 0 = N
#
# This is the population/signal-processing convention used in the lecture:
#     (1/N) * sum( (w[n] - sample_mean)^2 )
#
# If ddof=1 were used, NumPy would divide by N-1 instead.
variance_w1 = np.var(w1, ddof=0)
variance_w2 = np.var(w2, ddof=0)

# -------------------------------------------------------------------------
# PRINT SPECIFIED MODEL PARAMETERS AND MEASURED SAMPLE STATISTICS
# -------------------------------------------------------------------------
# The Gaussian model was explicitly constructed with zero mean.
print(f"Specified noise mean:       {0:.4f}")

# These are empirical means calculated from two different finite realizations.
print(f"Measured mean of w1:        {mean_w1:.4f}")
print(f"Measured mean of w2:        {mean_w2:.4f}")

# print() with no argument produces a blank line for readability.
print()

# This is the variance parameter we requested when generating the noise.
print(f"Specified noise variance:   {noise_variance:.4f}")

# These measured finite-record variances should be close to 0.05.
# They need not equal 0.05 exactly and need not equal each other.
print(f"Measured variance of w1:    {variance_w1:.4f}")
print(f"Measured variance of w2:    {variance_w2:.4f}")
````

## 5. Display the noise distribution

A histogram uses the samples from one finite realization to estimate the shape of the underlying probability density.

### Important distinction

The histogram is **not** the theoretical Gaussian PDF itself. It is an empirical estimate constructed from the 1000 values in `w1`.

`bins=30` divides the amplitude range into 30 histogram bins.

`density=True` normalizes the histogram so that its total area is approximately 1, making the vertical axis an estimated probability density rather than a raw sample count.

````python
# -------------------------------------------------------------------------
# SECTION 5: HISTOGRAM OF ONE NOISE REALIZATION
# -------------------------------------------------------------------------

# Create a new figure.
plt.figure()

# Build a histogram from the 1000 numerical values in w1.
#
# bins=30:
#     Divide the displayed amplitude range into 30 intervals.
#
# density=True:
#     Normalize the histogram so the total area is approximately 1.
#     The vertical axis is therefore an estimate of probability density,
#     not simply the number of samples in each bin.
#
# The histogram is an empirical estimate from ONE realization.
# It is not the exact analytical Gaussian probability-density function.
plt.hist(w1, bins=30, density=True)

plt.xlabel("Noise amplitude")
plt.ylabel("Estimated probability density")
plt.title("Histogram of One Gaussian-Noise Realization")

plt.show()
````

## 6. Add a constant offset

### Prediction

What should change when the same constant is added to every sample?

- Mean: changes.
- Variance: does not change.
- Average power: generally changes.

We create

\[
w_{	ext{offset}}[n]=w_1[n]+c.
\]

Adding \(c\) shifts every sample and therefore shifts the mean by \(c\). However, variance measures squared distance **from the mean**, so shifting both the samples and their mean by the same amount leaves the deviations unchanged.

Average power is different. For a real-valued signal,

\[
P = E[X^2].
\]

A nonzero mean contributes to average power. In fact,

\[
E[X^2]=\mathrm{Var}(X)+(E[X])^2.
\]

### C/C++ mental model

`w1 + c` applies the scalar `c` to every element of the NumPy array. Conceptually, it replaces an explicit loop such as `w_offset[i] = w1[i] + c`.

````python
# -------------------------------------------------------------------------
# SECTION 6: ADD A CONSTANT OFFSET
# -------------------------------------------------------------------------

# Constant offset to add to every noise sample.
c = 2

# NumPy "broadcasts" the scalar c across the entire array.
# Conceptually, this is equivalent to:
#
#     for (int i = 0; i < N; ++i)
#         w_offset[i] = w1[i] + c;
#
# No explicit Python loop is required.
w_offset = w1 + c

# -------------------------------------------------------------------------
# MEAN BEFORE AND AFTER THE SHIFT
# -------------------------------------------------------------------------
# Adding c to every sample should shift the mean by approximately c.
mean_before = np.mean(w1)
mean_after = np.mean(w_offset)

# -------------------------------------------------------------------------
# VARIANCE BEFORE AND AFTER THE SHIFT
# -------------------------------------------------------------------------
# Variance measures squared deviation from the mean.
#
# If Y = X + c, then
#     Y - E[Y] = (X+c) - (E[X]+c) = X - E[X]
#
# so adding a constant does not change variance.
variance_before = np.var(w1, ddof=0)
variance_after = np.var(w_offset, ddof=0)

# -------------------------------------------------------------------------
# AVERAGE POWER BEFORE AND AFTER THE SHIFT
# -------------------------------------------------------------------------
# Average power of a discrete record is the average squared magnitude:
#     P = (1/N) sum |x[n]|^2
#
# np.abs(w1) computes the magnitude of every sample.
# **2 squares every magnitude.
# np.mean(...) averages those squared magnitudes.
power_before = np.mean(np.abs(w1)**2)
power_after = np.mean(np.abs(w_offset)**2)

# -------------------------------------------------------------------------
# PRINT RESULTS
# -------------------------------------------------------------------------
print(f"Constant offset added: {c:.2f}")
print()

print(f"Mean before offset:       {mean_before:.4f}")
print(f"Mean after offset:        {mean_after:.4f}")
print()

# These two values should be essentially identical except for floating-point
# roundoff because a constant shift does not change spread around the mean.
print(f"Variance before offset:   {variance_before:.4f}")
print(f"Variance after offset:    {variance_after:.4f}")
print()

# Power generally changes because a DC/mean component contributes power.
print(f"Power before offset:      {power_before:.4f}")
print(f"Power after offset:       {power_after:.4f}")
````

## 7. Calculate average signal power

For a discrete-time record, the measured average power is

\[
P_x=
rac{1}{N}\sum_{n=0}^{N-1}|x[n]|^2.
\]

For a sinusoid

\[
x(t)=A\cos(2\pi f_0 t+\phi),
\]

the theoretical average power over an integer number of periods is

\[
P_x=
rac{A^2}{2}.
\]

This follows from

\[
\cos^2(	heta)=
rac{1+\cos(2	heta)}{2},
\]

because the cosine term averages to zero over complete periods, leaving an average value of \(1/2\).

Here \(A=1\), so the theoretical power is 0.5.

### Python syntax reminder

`A**2` means \(A^2\). In Python, `**` is exponentiation.

````python
# -------------------------------------------------------------------------
# SECTION 7: AVERAGE POWER OF THE DETERMINISTIC SINUSOID
# -------------------------------------------------------------------------

# Measured average power from the finite discrete-time record:
#
#     P = (1/N) sum |x1[n]|^2
#
# np.abs(x1) gives sample magnitudes.
# **2 squares them.
# np.mean(...) averages the squared magnitudes.
#
# x1 is real-valued here, so abs(x1)^2 is simply x1^2. Writing it this way
# also matches the general complex-signal power formula used later.
signal_power = np.mean(np.abs(x1)**2)

# The theoretical time-average power of
#     x(t) = A cos(2*pi*f0*t + phi)
# is
#     A^2 / 2.
#
# Reason:
#     cos^2(theta) = [1 + cos(2*theta)] / 2
# and the cosine term averages to zero over complete periods.
#
# In Python, ** is the exponentiation operator, so A**2 means A squared.
theoretical_signal_power = A**2 / 2

# The record contains an integer number of sinusoidal periods, so the measured
# result should match the theoretical 0.5 very closely.
print(f"Measured sinusoid power:     {signal_power:.4f}")
print(f"Theoretical sinusoid power:  {theoretical_signal_power:.4f}")
````

## 8. Add noise to the sinusoid

We now build the standard additive-noise model

\[
y[n]=x[n]+w[n].
\]

`w_low` is a **new realization** of zero-mean Gaussian noise with variance 0.05. It is not the same array as `w1`, even though both come from the same statistical model.

The measured powers are

\[
P_s=
rac{1}{N}\sum |x[n]|^2
\]

and

\[
P_n=
rac{1}{N}\sum |w[n]|^2.
\]

The linear SNR is

\[
\mathrm{SNR}=
rac{P_s}{P_n},
\]

and the decibel value is

\[
\mathrm{SNR}_{dB}=10\log_{10}(\mathrm{SNR}).
\]

### Teaching point

For zero-mean noise, average power and variance coincide theoretically:

\[
E[w^2]=\mathrm{Var}(w)
\]

when \(E[w]=0\).

Because this is a finite random realization, the measured noise power will be close to 0.05 rather than exactly 0.05.

````python
# -------------------------------------------------------------------------
# SECTION 8: ADD GAUSSIAN NOISE TO THE SINUSOID
# -------------------------------------------------------------------------
# Standard additive-noise communication model:
#
#     y[n] = x[n] + w[n]
#
# We intentionally generate a NEW realization named w_low.
# It has the same statistical model as w1 and w2:
#     mean = 0
#     variance = noise_variance = 0.05
# but it contains a different sequence of random samples.

# rng.standard_normal(N) gives unit-variance Gaussian samples.
# Multiplying by sqrt(noise_variance) gives the desired variance.
w_low = np.sqrt(noise_variance) * rng.standard_normal(N)

# Add noise sample-by-sample to the clean deterministic sinusoid.
# Both x1 and w_low have N elements, so NumPy performs elementwise addition.
y_low = x1 + w_low

# -------------------------------------------------------------------------
# MEASURE SIGNAL AND NOISE POWER FROM THESE PARTICULAR ARRAYS
# -------------------------------------------------------------------------
# Signal power:
#     P_s = average |x[n]|^2
measured_signal_power = np.mean(np.abs(x1)**2)

# Noise power:
#     P_n = average |w[n]|^2
#
# For ideal zero-mean noise, theoretical average power equals variance.
# For this finite realization, the measured value should be near 0.05.
measured_noise_power = np.mean(np.abs(w_low)**2)

# -------------------------------------------------------------------------
# COMPUTE SNR
# -------------------------------------------------------------------------
# Linear signal-to-noise ratio:
#     SNR = P_s / P_n
snr_linear_low = measured_signal_power / measured_noise_power

# Convert a POWER ratio to decibels using 10*log10(...).
# The factor is 10 because SNR is a power ratio.
snr_db_low = 10 * np.log10(snr_linear_low)

# -------------------------------------------------------------------------
# PRINT NUMERICAL RESULTS
# -------------------------------------------------------------------------
print(f"Signal power:           {measured_signal_power:.4f}")
print(f"Noise power:            {measured_noise_power:.4f}")
print(f"Linear SNR:             {snr_linear_low:.4f}")
print(f"SNR in dB:              {snr_db_low:.2f} dB")

# -------------------------------------------------------------------------
# COMPARE CLEAN AND NOISY SIGNALS IN THE TIME DOMAIN
# -------------------------------------------------------------------------
plt.figure()

# Plot only the first 100 samples, equal to 0.1 seconds.
plt.plot(t[:100], x1[:100], linewidth=1.5, label="Clean signal")
plt.plot(t[:100], y_low[:100], label="Signal plus noise")

plt.xlabel("Time (seconds)")
plt.ylabel("Amplitude")
plt.title("Clean and Noisy Sinusoid")

# legend() uses the label strings specified in the two plot calls above.
plt.legend()

plt.show()
````

## 9. Increase the noise variance

### Prediction

If signal power stays fixed while noise power increases, what happens to SNR?

We increase the specified noise variance from

\[
0.05 
ightarrow 0.20,
\]

which is a factor of four.

Since

\[
\mathrm{SNR}=
rac{P_s}{P_n},
\]

a fourfold increase in noise power should reduce SNR by approximately

\[
10\log_{10}(4)pprox 6.02	ext{ dB}.
\]

The measured change will not necessarily be exactly 6.02 dB because the low-noise and high-noise arrays are independent finite realizations.

### Plotting note

The two panels show the same clean sinusoid against two different received signals. Matplotlib determines the plot limits from the data unless common limits are explicitly requested. The code below is left exactly as written in the original notebook.

````python
# -------------------------------------------------------------------------
# SECTION 9: INCREASE THE NOISE VARIANCE
# -------------------------------------------------------------------------
# We now hold the clean signal x1 fixed while increasing the noise variance.
#
# Original noise variance: 0.05
# New noise variance:      0.20
#
# The specified noise variance increases by a factor of 4.
# If noise power increased by exactly 4 while signal power remained fixed,
# SNR would decrease by
#     10*log10(4) ~= 6.02 dB.
#
# Because these are finite independent noise realizations, the measured SNR
# change will generally be close to, but not exactly, -6.02 dB.

# Desired variance for the stronger-noise case.
strong_noise_variance = 0.20

# Generate a new Gaussian-noise realization with variance 0.20.
w_high = np.sqrt(strong_noise_variance) * rng.standard_normal(N)

# Add the stronger noise realization to exactly the same clean signal x1.
y_high = x1 + w_high

# Measure the average power of this finite high-noise realization.
high_noise_power = np.mean(np.abs(w_high)**2)

# Signal power is still measured_signal_power from Section 8 because x1 has
# not changed.
#
# Compute the new linear SNR using the higher measured noise power.
snr_linear_high = measured_signal_power / high_noise_power

# Convert the higher-noise SNR to dB.
snr_db_high = 10 * np.log10(snr_linear_high)

# -------------------------------------------------------------------------
# PRINT LOW-NOISE CASE
# -------------------------------------------------------------------------
print("LOW-NOISE CASE")

# This is the MODEL variance used to generate w_low.
print(f"Noise variance:         {noise_variance:.4f}")

# This is the FINITE-REALIZATION power measured from w_low.
print(f"Measured noise power:   {measured_noise_power:.4f}")

# This is the SNR calculated from the finite arrays.
print(f"Measured SNR:           {snr_db_low:.2f} dB")
print()

# -------------------------------------------------------------------------
# PRINT HIGH-NOISE CASE
# -------------------------------------------------------------------------
print("HIGH-NOISE CASE")

# This is the MODEL variance used to generate w_high.
print(f"Noise variance:         {strong_noise_variance:.4f}")

# This is the FINITE-REALIZATION power measured from w_high.
print(f"Measured noise power:   {high_noise_power:.4f}")

# This is the resulting measured SNR.
print(f"Measured SNR:           {snr_db_high:.2f} dB")
print()

# Difference in dB between the two measured SNR values.
# Expect something close to -6.02 dB, not necessarily exactly -6.02 dB.
print(f"Change in SNR:          {snr_db_high - snr_db_low:.2f} dB")

# -------------------------------------------------------------------------
# PLOT THE LOW- AND HIGH-NOISE CASES
# -------------------------------------------------------------------------
# Create two vertically stacked axes that share the same x-axis scale.
#
# NOTE: This line is intentionally preserved exactly from the original script.
# sharex=True synchronizes the horizontal scale. The original code does not
# explicitly force the two y-axis limits to be identical.
fig, ax = plt.subplots(2, 1, figsize=(8, 5), sharex=True)

# ----- Upper panel: lower noise -----
ax[0].plot(t[:100], x1[:100], linewidth=1.5, label="Clean signal")
ax[0].plot(t[:100], y_low[:100], label="Received signal")
ax[0].set_ylabel("Amplitude")
ax[0].set_title(f"Lower Noise: SNR = {snr_db_low:.2f} dB")
ax[0].legend()

# ----- Lower panel: higher noise -----
ax[1].plot(t[:100], x1[:100], linewidth=1.5, label="Clean signal")
ax[1].plot(t[:100], y_high[:100], label="Received signal")
ax[1].set_xlabel("Time (seconds)")
ax[1].set_ylabel("Amplitude")
ax[1].set_title(f"Higher Noise: SNR = {snr_db_high:.2f} dB")
ax[1].legend()

# Adjust subplot spacing so labels and titles fit cleanly.
plt.tight_layout()

# Render the complete two-panel figure.
plt.show()
````

## Week 1 summary

1. A deterministic sinusoid repeats exactly when all parameters are unchanged.
2. Noise records from the same model are different realizations.
3. Sample mean estimates the center of a finite record.
4. With `ddof=0`, NumPy variance divides the squared-deviation sum by `N`.
5. Average signal power is the average squared magnitude.
6. A sinusoid of amplitude \(A\) has theoretical average power \(A^2/2\).
7. Adding a constant changes the mean and generally the power, but not the variance.
8. `w_low` and `w_high` are fresh Gaussian-noise realizations used to form received signals.
9. Linear SNR is signal power divided by noise power.
10. SNR in dB is \(10\log_{10}(\mathrm{SNR})\).
11. Increasing noise power while holding signal power fixed reduces SNR.
