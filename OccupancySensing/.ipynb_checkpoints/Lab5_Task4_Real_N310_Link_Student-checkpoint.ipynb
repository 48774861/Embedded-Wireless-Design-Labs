## Finite-file and capture calculations


The sample rate of 1 MS/s, center frequency of 2.45 GHz, period length of 8192 samples, and one warm-up period were kept the same as Lab 5 to preserve the validated OFDM waveform, synchronization process, and channel-estimation receiver without requiring modification of the underlying signal-processing chain. The repeat count was increased from 13 to 50 periods to provide 49 usable channel estimates per capture, improving the statistical robustness of the occupancy-sensing analysis by allowing channel estimates to be averaged and compared across many observations. The acquisition lead was reduced from 2,000,000 samples to 250,000 samples (2 s to 0.25 s) because only a short delay is needed to ensure the receiver is recording before transmission begins. Similarly, the capture duration was reduced from 15 s to 5 s since the transmitted OFDM burst occupies less than one second, significantly reducing file size and data-processing requirements while still providing ample margin to capture the complete transmission.

```python
sample_rate_sps = 1_000_000
period_samples = 8192
repeat_count = 50
warmup_periods = 1
acquisition_lead_samples = 250_000
capture_seconds = 5.0
center_frequency_hz = 2_450_000_000.0

# Exact counts
payload_burst_samples = repeat_count * period_samples
transmit_samples = acquisition_lead_samples + payload_burst_samples
transmit_bytes = transmit_samples * 8          # complex64

capture_samples = int(capture_seconds * sample_rate_sps)
capture_bytes = capture_samples * 8            # complex64

expected_frames = repeat_count - warmup_periods

print("OFDM burst samples:", payload_burst_samples)
print("guarded TX samples and bytes:", transmit_samples, transmit_bytes)
print("RX samples and bytes:", capture_samples, capture_bytes)
print("graded frames:", expected_frames)
```

```python
import numpy as np
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))

# 0 people runs
for run in range(1, 6):
    H = results[(0, run)]["channel_estimates"]
    mags = np.abs(H)

    subcarrier_std = np.std(mags, axis=0)

    plt.plot(
        subcarrier_std,
        color="blue",
        alpha=0.5,
        label="0 People" if run == 1 else None
    )

# 12 people runs
for run in range(1, 11):
    H = results[(12, run)]["channel_estimates"]
    mags = np.abs(H)

    subcarrier_std = np.std(mags, axis=0)

    plt.plot(
        subcarrier_std,
        color="red",
        alpha=0.5,
        label="12 People" if run == 1 else None
    )

plt.xlabel("Subcarrier Index")
plt.ylabel("Magnitude Std Dev")
plt.title("Subcarrier Variability Across Runs")
plt.grid(True, alpha=0.3)
plt.legend()

plt.tight_layout()
plt.savefig("subcarrier_variability_all_runs.png", dpi=300)
plt.close()
```

```python
zero_runs = []
twelve_runs = []

for run in range(1, 11):

    twelve_runs.append(
        np.std(np.abs(results[(12, run)]["channel_estimates"]), axis=0)
    )
for run in range(1, 6):
    zero_runs.append(
        np.std(np.abs(results[(0, run)]["channel_estimates"]), axis=0)
    )

zero_runs = np.array(zero_runs)
twelve_runs = np.array(twelve_runs)

zero_mean = np.mean(zero_runs, axis=0)
zero_std = np.std(zero_runs, axis=0)

twelve_mean = np.mean(twelve_runs, axis=0)
twelve_std = np.std(twelve_runs, axis=0)

plt.figure(figsize=(12,6))

plt.plot(zero_mean, color="blue", label="0 People")
plt.fill_between(
    np.arange(52),
    zero_mean - zero_std,
    zero_mean + zero_std,
    color="blue",
    alpha=0.2
)

plt.plot(twelve_mean, color="red", label="12 People")
plt.fill_between(
    np.arange(52),
    twelve_mean - twelve_std,
    twelve_mean + twelve_std,
    color="red",
    alpha=0.2
)

plt.xlabel("Subcarrier Index")
plt.ylabel("Magnitude Std Dev")
plt.title("Mean Subcarrier Variability Across Runs")
plt.legend()
plt.grid(alpha=0.3)

plt.tight_layout()
plt.savefig("subcarrier_variability_summary.png", dpi=300)
plt.close()
```

Looking at the plot:

The new 12-person measurements (red) collected 2 hours later fall right on top of the original 12-person measurements.
There isn't an obvious second cluster corresponding to "old 12-person" vs "new 12-person."
The 12-person traces remain consistently below the 0-person traces.
The separation appears larger than the spread within the 12-person measurements themselves.

That is not what I'd expect if the effect were primarily random phase drift over time.

Mean channel phase showed a statistically significant difference between the 0-person and 12-person measurements (Welch's t-test, p = 0.0018). However, the phase distributions partially overlapped, indicating that mean phase alone is not sufficient for perfect occupancy classification. The observed result suggests that average phase contains occupancy-related information, whereas channel magnitude and correlation metrics did not show statistically significant occupancy-dependent behavior.

```python
def plot_correlation(results, output_dir):

    H_ref = results[(0,1)]["mean_channel_estimate"]

    fig, ax = plt.subplots(figsize=(8,5))

    for (people, run), result in sorted(results.items()):

        H = result["mean_channel_estimate"]

        corr = channel_corr(H_ref, H)

        ax.scatter(
            people,
            corr,
            s=60,
            label=f"run {run}" if people == 0 else None
        )

    ax.set(
        xlabel="Number of People",
        ylabel="Channel Correlation",
        title="Correlation Relative to Empty-Room Run 1"
    )

    ax.grid(alpha=0.25)

    save_figure(fig, output_dir, "occupancy_correlation")
```

```python
def plot_nmse(results, output_dir):

    H_ref = results[(0,1)]["mean_channel_estimate"]

    fig, ax = plt.subplots(figsize=(8,5))

    for (people, run), result in sorted(results.items()):

        H = result["mean_channel_estimate"]

        nmse = channel_nmse(H_ref, H)

        ax.scatter(
            people,
            nmse,
            s=60,
        )

    ax.set(
        xlabel="Number of People",
        ylabel="NMSE",
        title="NMSE Relative to Empty-Room Run 1"
    )

    ax.grid(alpha=0.25)

    save_figure(fig, output_dir, "occupancy_nmse")
```

```python
def plot_mean_correlation(results, output_dir):

    H_ref = results[(0,1)]["mean_channel_estimate"]

    occupancy = sorted(
        set(p for p, r in results.keys())
    )

    means = []
    stds = []

    for p in occupancy:

        vals = []

        for run in [1,2,3]:

            if (p,run) not in results:
                continue

            H = results[(p,run)]["mean_channel_estimate"]

            vals.append(
                channel_corr(H_ref, H)
            )

        means.append(np.mean(vals))
        stds.append(np.std(vals))

    fig, ax = plt.subplots(figsize=(8,5))

    ax.errorbar(
        occupancy,
        means,
        yerr=stds,
        fmt='o-',
        capsize=5
    )

    ax.set(
        xlabel="Number of People",
        ylabel="Mean Correlation",
        title="Occupancy Impact on Channel Correlation"
    )

    ax.grid(alpha=0.25)

    save_figure(fig, output_dir, "occupancy_mean_correlation")
```
