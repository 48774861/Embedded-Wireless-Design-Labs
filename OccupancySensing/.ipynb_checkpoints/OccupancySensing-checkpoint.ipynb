## Finite-file and capture calculations


The sample rate of 1 MS/s, center frequency of 2.45 GHz, period length of 8192 samples, and one warm-up period were kept the same as Lab 5 to preserve the validated OFDM waveform, synchronization process, and channel-estimation receiver without requiring modification of the underlying signal-processing chain. The repeat count was increased from 13 to 50 periods to provide 49 usable channel estimates per capture, improving the statistical robustness of the occupancy-sensing analysis by allowing channel estimates to be averaged and compared across many observations. The acquisition lead was reduced from 2,000,000 samples to 250,000 samples (2 s to 0.25 s) because only a short delay is needed to ensure the receiver is recording before transmission begins. Similarly, the capture duration was reduced from 15 s to 5 s since the transmitted OFDM burst occupies less than one second, significantly reducing file size and data-processing requirements while still providing ample margin to capture the complete transmission.

```python
sample_rate_sps = 1_000_000
period_samples = 8192
repeat_count = 50
warmup_periods = 1
acquisition_lead_samples = 250_000
capture_seconds = 5.0
center_frequency_hz = 2_450_000_000.0

# Exact counts
payload_burst_samples = repeat_count * period_sçamples
transmit_samples = acquisition_lead_samples + payload_burst_samples
transmit_bytes = transmit_samples * 8          # complex64

capture_samples = int(capture_seconds * sample_rate_sps)
capture_bytes = capture_samples * 8            # complex64

expected_frames = repeat_count - warmup_periods

print("OFDM burst samples:", payload_burst_samples)
print("guarded TX samples and bytes:", transmit_samples, transmit_bytes)
print("RX samples and bytes:", capture_samples, capture_bytes)
print("graded frames:", expected_frames)
```

### Channel Magnitude Plot

```python
zero_runs = []
twelve_runs = []

for run in range(1, 11):

    H_mean = np.mean(
        results[(0, run)]["channel_estimates"],
        axis=0
    )

    zero_runs.append(
        20*np.log10(
            np.maximum(np.abs(H_mean), 1e-12)
        )
    )

for run in range(1, 11):
    H_mean = np.mean(
        results[(12, run)]["channel_estimates"],
        axis=0
    )

    twelve_runs.append(
        20*np.log10(
            np.maximum(np.abs(H_mean), 1e-12)
        )
    )

zero_runs = np.array(zero_runs)
twelve_runs = np.array(twelve_runs)

zero_mean = np.mean(zero_runs, axis=0)
zero_std  = np.std(zero_runs, axis=0)

twelve_mean = np.mean(twelve_runs, axis=0)
twelve_std  = np.std(twelve_runs, axis=0)

print("\nMagnitude Variability Comparison")
print(f"0 People  mean std-dev: {np.mean(zero_std):.3f} dB")
print(f"12 People mean std-dev: {np.mean(twelve_std):.3f} dB")

print("\nAverage Magnitude Comparison")
print(f"0 People  mean magnitude: {np.mean(zero_mean):.3f} dB")
print(f"12 People mean magnitude: {np.mean(twelve_mean):.3f} dB")

plt.figure(figsize=(12, 6))

plt.plot(
    zero_mean,
    color="blue",
    label="0 People"
)

plt.fill_between(
    np.arange(52),
    zero_mean - zero_std,
    zero_mean + zero_std,
    color="blue",
    alpha=0.2
)

plt.plot(
    twelve_mean,
    color="red",
    label="12 People"
)

plt.fill_between(
    np.arange(52),
    twelve_mean - twelve_std,
    twelve_mean + twelve_std,
    color="red",
    alpha=0.2
)

plt.xlabel("Subcarrier Index")
plt.ylabel("Channel Magnitude (dB)")
plt.title("Mean OFDM Channel Magnitude Across Runs")
plt.legend()
plt.grid(alpha=0.3)

plt.tight_layout()
plt.savefig(
    "work/figures/subcarrier_variability_db_summary.png",
    dpi=300
)
plt.close()

display(
    Image(
        filename="work/figures/subcarrier_variability_db_summary.png"
    )
)
```

### Channel Phase Plot

```python
zero_runs = []
twelve_runs = []

for run in range(1,11):
    H_mean = np.mean(
        results[(0, run)]["channel_estimates"],
        axis=0
    )
    zero_runs.append(
        np.unwrap(np.angle(H_mean))
    )

for run in range(1, 11):
    H_mean = np.mean(
        results[(12, run)]["channel_estimates"],
        axis=0
    )
    twelve_runs.append(
        np.unwrap(np.angle(H_mean))
    )

zero_runs = np.array(zero_runs)
twelve_runs = np.array(twelve_runs)

zero_mean = np.mean(zero_runs, axis=0)
zero_std = np.std(zero_runs, axis=0)

twelve_mean = np.mean(twelve_runs, axis=0)
twelve_std = np.std(twelve_runs, axis=0)

print("\nPhase Variability Comparison")
print(f"0 People  mean std-dev: {np.mean(zero_std):.3f} rad")
print(f"12 People mean std-dev: {np.mean(twelve_std):.3f} rad")

print("\nAverage Phase Comparison")
print(f"0 People  mean magnitude: {np.mean(zero_mean):.3f} rad")
print(f"12 People mean magnitude: {np.mean(twelve_mean):.3f} rad")

plt.figure(figsize=(12,6))

plt.plot(zero_mean, color="blue", label="0 People")
plt.fill_between(
    np.arange(52),
    zero_mean - zero_std,
    zero_mean + zero_std,
    color="blue",
    alpha=0.2
)

plt.plot(twelve_mean, color="red", label="12 People")
plt.fill_between(
    np.arange(52),
    twelve_mean - twelve_std,
    twelve_mean + twelve_std,
    color="red",
    alpha=0.2
)

plt.xlabel("Subcarrier Index")
plt.ylabel("Phase (rad)")
plt.title("Mean Subcarrier Phase Variability Across Runs")
plt.legend()
plt.grid(alpha=0.3)

plt.tight_layout()
plt.savefig(
    "work/figures/subcarrier_phase_variability_summary.png",
    dpi=300
)
plt.close()

display(
    Image(
        filename="work/figures/subcarrier_phase_variability_summary.png"
    )
)
```
